```python
import cv2
import json
import numpy as np
import tkinter as tk
from tkinter import filedialog
from tensorflow.keras.models import load_model
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input

model = load_model("flower_classifier.keras")

with open("class_names.json", "r") as f:
    class_names = json.load(f)

IMG_SIZE = 160

root = tk.Tk()
root.withdraw()
root.attributes('-topmost', True)
image_path = filedialog.askopenfilename(
    parent=root,
    title="Select a flower image",
    filetypes=[("Image files", "*.jpg *.jpeg *.png *.bmp")]
)
root.destroy()

if not image_path:
    print("No image selected.")
else:
    image = cv2.imread(image_path)
    resized = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
    rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)
    array = preprocess_input(rgb.astype("float32"))
    array = np.expand_dims(array, axis=0)

    prediction = model.predict(array)
    predicted_idx = np.argmax(prediction)
    predicted_class = class_names[predicted_idx]
    confidence = prediction[0][predicted_idx] * 100

    display = cv2.resize(image, (500, 500))
    label = f"{predicted_class} ({confidence:.1f}%)"
    cv2.putText(display, label, (20, 40), cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 255, 0), 2)
    cv2.imshow("Flower Classifier", display)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
